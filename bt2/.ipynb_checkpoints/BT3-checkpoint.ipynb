```python
# 3.1 — nạp dữ liệu bằng ĐƯỜNG DẪN TƯƠNG ĐỐI
from pathlib import Path
import pandas as pd

# Thử vài vị trí hợp lý để notebook chạy được cả khi mở từ thư mục cha.
UNG_VIEN = [
    Path("../lab_data/sinhvien.csv"),   # notebook nằm trong lab_exercises/  <- đúng chuẩn
    Path("lab_data/sinhvien.csv"),      # notebook nằm ở thư mục gốc học phần
    Path("sinhvien.csv"),               # CSV nằm cùng thư mục với notebook
]
CSV_PATH = next((p for p in UNG_VIEN if p.exists()), None)

if CSV_PATH is None:
    raise FileNotFoundError(
        "Không tìm thấy sinhvien.csv.\n"
        "Notebook phải nằm trong thư mục lab_exercises/, CẠNH thư mục lab_data/.\n"
        "Thư mục hiện tại : " + str(Path.cwd()) + "\n"
        "Đã thử các đường dẫn: " + ", ".join(str(p) for p in UNG_VIEN)
    )

df = pd.read_csv(CSV_PATH)
print("Đã nạp:", CSV_PATH.resolve())
df.head()
```

```python
# 3.2 — MÔ TẢ dữ liệu trước khi động vào nó
print("Số dòng:", df.shape[0])
print("Số cột :", df.shape[1])
print("Tên cột:", list(df.columns))

print("\n--- Kiểu dữ liệu từng cột ---")
print(df.dtypes)

# TODO: in SỐ giá trị thiếu của từng cột.      Gợi ý: df.isna().sum()
# TODO: in thêm TỈ LỆ thiếu theo %, làm tròn 2 chữ số.
#       Gợi ý: df.isna().mean() * 100  -> rồi .round(2)
print("\n--- Số giá trị thiếu ---")
print(df.isna().sum())

print("\n--- Tỉ lệ thiếu (%) ---")
print((df.isna().mean() * 100).round(2))
```

```python
# 3.3a — MỘT LỖI KIỂU DỮ LIỆU ĐANG NẰM NGAY ĐÂY
print("dtype của mssv khi read_csv đọc mặc định:", df["mssv"].dtype)
print("Một giá trị     :", df["mssv"].iloc[0], "->", type(df["mssv"].iloc[0]).__name__)
print()
print("Trung bình cộng MSSV:", df["mssv"].mean())
print("MSSV lớn nhất       :", df["mssv"].max())
print()
print("Hai con số trên tính được, nhưng KHÔNG CÓ NGHĨA GÌ.")
print("MSSV là ĐỊNH DANH: không cộng, không trừ, không lấy trung bình.")
```

```python
# 3.3b — SỬA: ép mssv thành chuỗi ngay khi đọc
# TODO: nạp lại df từ CSV_PATH, thêm tham số dtype={"mssv": str}
df = pd.read_csv(CSV_PATH, dtype={"mssv": str})

print("dtype sau khi sửa:", df["mssv"].dtype)
print("Một giá trị       :", repr(df["mssv"].iloc[0]))
print("shape             :", df.shape)
```

```python
# 3.4 — tự kiểm tra sau khi sửa
assert isinstance(df["mssv"].iloc[0], str), "mssv phải là chuỗi, không phải số"
assert df["mssv"].iloc[0] == "2200001", "Giá trị đầu tiên phải là chuỗi '2200001'"
assert df.shape == (1200, 9), f"Dữ liệu phải có 1200 dòng x 9 cột, đang là {df.shape}"
assert df["mssv"].duplicated().sum() == 0, "MSSV phải là duy nhất"
assert df["gpa"].isna().sum() == 55, "Phải có đúng 55 sinh viên thiếu GPA"

print("3.3 OK — mssv giờ là định danh dạng chuỗi; 1200 mã, không trùng.")
```
