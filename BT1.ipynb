```python
import sys

print("Python version :", sys.version.split()[0])
print("Interpreter    :", sys.executable)

print("Major          :", sys.version_info.major)
print("Minor          :", sys.version_info.minor)
print("Python >= 3.11 :", (sys.version_info.major, sys.version_info.minor) >= (3, 11))
```

```python
LIBS = {
    "numpy":        "numpy",
    "pandas":       "pandas",
    "matplotlib":   "matplotlib",
    "seaborn":      "seaborn",
    "scikit-learn": "sklearn",
}

import numpy as np
import pandas as pd

print("numpy       :", np.__version__)
print("pandas      :", pd.__version__)
```

```python
import matplotlib
import seaborn
import sklearn

print("matplotlib  :", matplotlib.__version__)
print("seaborn     :", seaborn.__version__)
print("scikit-learn:", sklearn.__version__)
```

```python
import importlib

versions = {}

for pip_name, import_name in LIBS.items():
    try:
        mod = importlib.import_module(import_name)
        versions[pip_name] = getattr(mod, "__version__", "?")
        print(f"PASS  {pip_name:<14} {versions[pip_name]}")
    except ImportError:
        versions[pip_name] = None
        print(f"FAIL  {pip_name:<14} chưa cài  ->  pip install {pip_name}")

print()
print("Số thư viện PASS:", sum(v is not None for v in versions.values()), "/", len(LIBS))
```

```python
thieu = [name for name, v in versions.items() if v is None]
assert not thieu, "Thiếu thư viện: " + ", ".join(thieu) + "  ->  pip install " + " ".join(thieu)
assert sys.version_info >= (3, 11), "Cần Python >= 3.11, đang chạy " + sys.version.split()[0]

print("Bài 1 OK — môi trường đã sẵn sàng.")
```
