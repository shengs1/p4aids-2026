```python
# 4.5 — kiểm tra trạng thái Git. Ô này chỉ ĐỌC, không thay đổi gì.
import shutil
import subprocess


def chay(*lenh):
    r = subprocess.run(lenh, capture_output=True, text=True)
    return r.returncode, (r.stdout + r.stderr).strip()


if shutil.which("git") is None:
    print("FAIL  git chưa được cài. Tải tại https://git-scm.com/downloads")
else:
    print("PASS  ", chay("git", "--version")[1])

    ma, ket_qua = chay("git", "rev-parse", "--is-inside-work-tree")
    if ma == 0 and ket_qua == "true":
        print("PASS   Thư mục này đã là một Git repository.")

        log = chay("git", "log", "--oneline", "-n", "3")[1]
        print("PASS   Commit gần nhất:" if log else "CHƯA   Chưa có commit nào — chạy git commit")
        print(log or "       (trống)")

        remote = chay("git", "remote", "-v")[1]
        print("\nRemote:")
        print(remote or "       (chưa có remote — chạy git remote add origin <url>)")
    else:
        print("CHƯA   Chưa phải Git repository.")
        print("       Mở Terminal tại thư mục lab_exercises/ rồi chạy: git init")
```
